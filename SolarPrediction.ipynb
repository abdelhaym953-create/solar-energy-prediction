# Feature Engineering

```python
df
```

```python
 #  Combine date + time
df["datetime"] = pd.to_datetime(df["data"].str.split().str[0] + " " + df["time"],format="%m/%d/%Y %H:%M:%S",errors="coerce")
```

```python
df["month"] = df["datetime"].dt.month
```

```python
# 3) Time features
df["hour"] = df["datetime"].dt.hour
df["minute"] = df["datetime"].dt.minute
df["second"] = df["datetime"].dt.second
```

```python
 # 4) Sunrise / Sunset parsing
df["sunrise_dt"] = pd.to_datetime(df["timesunrise"], format="%H:%M:%S", errors="coerce")
df["sunset_dt"] = pd.to_datetime(df["timesunset"], format="%H:%M:%S", errors="coerce")
```

```python
df["sunrise_minutes"] = ( df["sunrise_dt"].dt.hour * 60 + df["sunrise_dt"].dt.minute + df["sunrise_dt"].dt.second / 60)
##

df["sunset_minutes"] = ( df["sunset_dt"].dt.hour * 60 + df["sunset_dt"].dt.minute + df["sunset_dt"].dt.second / 60)
```

```python
# current time in minutes from midnight
df["current_minutes"] = df["hour"] * 60 + df["minute"] + df["second"] / 60
```

```python
 # relative to sunrise/sunset

df["minutes_since_sunrise"] = df["current_minutes"] - df["sunrise_minutes"]
df["minutes_to_sunset"] = df["sunset_minutes"] - df["current_minutes"]
```

```python
# To show the whole columns
pd.set_option('display.max_columns', None)
```

```python
# drop unnecessary columns
df.drop(columns=["unixtime", "data", "time", "datetime", "sunrise_dt", "sunset_dt", "current_minutes",'timesunrise','timesunset'],axis=1,inplace=True)
```

# Pre-For Machine Learning

```python
df
```

```python
# slit the data into features and target
x = df.drop('radiation',axis=1)
y = df['radiation']
```

```python
num_cols = x.select_dtypes(include='number').columns
```

```python
# numerical pipeline
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import RobustScaler
num_pipeline = Pipeline([
    ('Scaler', RobustScaler()) ])
num_pipeline
```

```python
# apply the numerical pipeline to the numerical columns
from sklearn.compose import ColumnTransformer
preprocessing = ColumnTransformer([('Numerical Pipeline', num_pipeline, num_cols) ], remainder='passthrough')
preprocessing
```

```python
import warnings
warnings.filterwarnings("ignore")
```

```python
from sklearn.model_selection import cross_validate
from sklearn.linear_model import LinearRegression
from sklearn.neighbors import KNeighborsRegressor
from sklearn.svm import SVR
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
#from sklearn.svm import SVR
from catboost import CatBoostRegressor
from xgboost import XGBRegressor
from lightgbm import LGBMRegressor
#-----------------------------------------------------------------------
models = [
    ('Linear Regression', LinearRegression()),
    ('KNN Regressor', KNeighborsRegressor()),
    ('Decision Tree Regressor', DecisionTreeRegressor(random_state=42)),
    ('Random Forest Regressor', RandomForestRegressor(random_state=42)),
   # ('Support Vector Regressor', SVR()),
    ('CatBoost Regressor', CatBoostRegressor( random_state=42, verbose=0)),
    ('XGBoost Regressor', XGBRegressor(random_state=42, verbosity=0)),
    ('LightGBM Regressor', LGBMRegressor(random_state=42, verbose=-1))
]
#-----------------------------------------------------------------------
for model in models:
        model_pipline = Pipeline(steps=[('Preprocessing',preprocessing),('Model',model[1])])
        result = cross_validate(model_pipline, x, y, cv=5 , scoring=['r2'], return_train_score=True)
        print(model[0])
        print('Train R2:', np.mean(result['train_r2']))
        print('Test R2:', np.mean(result['test_r2']).round(4))
        print('_'*50)
        


```

```python
from sklearn.model_selection import RandomizedSearchCV
from lightgbm import LGBMRegressor
para_grid = {
    'Model__n_estimators': [100, 200, 300,500, 800],
    'Model__learning_rate': [0.01, 0.03,0.05, 0.1],
    'Model__max_depth': [3,4, 5,6, 7],
    'Model__num_leaves': [20, 30, 40]
}
model_pipline = Pipeline(steps=[('Preprocessing',preprocessing),('Model',LGBMRegressor(random_state=42))])
random_search = RandomizedSearchCV(estimator=model_pipline, param_distributions=para_grid, n_iter=10, cv=5, scoring='r2', n_jobs=-1, random_state=42)

random_search.fit(x, y)

print("Best Score:", random_search.best_score_)
print("Best Params:", random_search.best_params_)

```

```python
best_model = random_search.best_estimator_
lgbm_model = best_model.named_steps['Model']
import pandas as pd

feature_importance = pd.DataFrame({
    'Feature': best_model.named_steps['Preprocessing'].get_feature_names_out(),
    'Importance': lgbm_model.feature_importances_
})

feature_importance = feature_importance.sort_values(
    by='Importance',
    ascending=False
)

print(feature_importance)
```

```python
px.bar(feature_importance, x='Feature', y='Importance', color_discrete_sequence=['blue'], text_auto=True)
```

```python
final_pipeline = Pipeline(steps=[('Preprocessing',preprocessing),('Model',LGBMRegressor(random_state=42, n_estimators=200, max_depth=5, num_leaves=30, learning_rate=0.03))])
final_pipeline.fit(x, y)
```
